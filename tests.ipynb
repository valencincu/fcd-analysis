```python
import os
from pydata.image import *
from pyfcd.fcd import *

reference_image_path  = os.path.join("examples", "static_images", "reference_2.png")
reference_image       = Image(reference_image_path)
displaced_images_path = os.path.join("examples", "propagating_waves", "202406_1457001663.bmp")
displaced_image       = Image(displaced_images_path)

displaced_image.select_roi()
reference_image.roi = displaced_image.roi

reference_image.show()
displaced_image.show()
```

```python
fcd = FCD(reference_image.processed, square_size=0.0022, height=0.0323625)
height_field = fcd.analyze(displaced_image.processed, full_output=True)

height_field.show()
```
